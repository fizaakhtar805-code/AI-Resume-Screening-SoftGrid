```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x="shortlisted", data=df)

plt.title("Shortlisted Candidates")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["years_experience"], bins=16)

plt.title("Years of Experience Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["skills_match_score"], bins=20)

plt.title("Skills Match Score Distribution")

plt.show()
```

```python
sns.countplot(x="education_level", data=df)

plt.title("Education Level Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df["education_level"] = le.fit_transform(df["education_level"])

df.head()
```

```python
df.to_csv("cleaned_resume_screening.csv", index=False)

print("Dataset saved successfully")
```
